```python
a = {
"Employee_ID":[101,102,103,104,105],
"Name":["Alice","Bob","Charlie","Diana","Edward"],
"Department":["HR","IT","IT","Marketing","Sales"],
"Age":[29,34,41,28,38],
"Salary":[50000,70000,65000,55000,60000],
"Years_of_Experience":[4,8,10,3,12],
"Joining_Date":["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
"Gender":["Female","Male","Male","Female","Male"],
"Bonus":[5000,7000,6000,4500,5000],
"Rating":[4.5,4.0,3.8,4.7,3.5]
}

df = pd.DataFrame(a)

print("Shape:",df.shape)
print("\nData info:")
print(df.info())

print("\nStatistics:")
print(df.describe())

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 3 rows:")
print(df.tail(3))

print("\nAverage salary:",df["Salary"].mean())
print("Total bonus:",df["Bonus"].sum())
print("Youngest age:",df["Age"].min())
print("Highest rating:",df["Rating"].max())

df = df.sort_values("Salary",ascending=False)

print("\nSorted by salary:")
print(df[["Name","Salary"]])

df["Performance"] = "Average"
df.loc[df["Rating"] >= 4.0,"Performance"] = "Good"
df.loc[df["Rating"] >= 4.5,"Performance"] = "Excellent"

print("\nPerformance:")
print(df[["Name","Rating","Performance"]])

print("\nMissing values:")
print(df.isnull().sum())
df = df.rename(columns={"Employee_ID":"ID"})
print("\nMore than 5 years experience:")
print(df[df["Years_of_Experience"] > 5][["Name","Years_of_Experience"]])
print("\nIT employees:")
print(df[df["Department"] == "IT"][["Name","Department"]])
df["Tax"] = df["Salary"] * 0.10

print("\nFinal data:")
print(df)
df.to_csv("employees_modified.csv",index=False)
print("\nSaved as employees_modified.csv")
```
