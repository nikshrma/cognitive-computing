```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3])

values, counts = np.unique(y, return_counts=True)

most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

```python
arr = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
Nikhil = np.linspace(10, 100, 25)

print("Array:", Nikhil)
print("Dimensions:", Nikhil.ndim)
print("Shape:", Nikhil.shape)
print("Total elements:", Nikhil.size)
print("Data type:", Nikhil.dtype)
print("Total bytes:", Nikhil.nbytes)

Nikhil_transpose = Nikhil.reshape(25, 1)

print("Transpose using reshape():")
print(Nikhil_transpose)

print("Can we use T attribute?", Nikhil.T)
```

```python
ucs420_Nikhil = np.array([[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]])

print("Original array:")
print(ucs420_Nikhil)

print("Mean:", np.mean(ucs420_Nikhil))
print("Median:", np.median(ucs420_Nikhil))
print("Maximum:", np.max(ucs420_Nikhil))
print("Minimum:", np.min(ucs420_Nikhil))
print("Unique elements:", np.unique(ucs420_Nikhil))

reshaped_ucs420_Nikhil = ucs420_Nikhil.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_Nikhil)

resized_ucs420_Nikhil = np.resize(ucs420_Nikhil, (2, 3))

print("Resized array:")
print(resized_ucs420_Nikhil)
```
