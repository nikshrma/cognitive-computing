## **Functions**

```python
def isPrime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number: "))

if isPrime(num):
    print(num, "is Prime")
else:
    print(num, "is Not Prime")
```
