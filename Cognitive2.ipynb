```python
n = 1024170220
L = []
while n!=0:
  dig = n%10
  L.append(dig*10)
  n=n//10
print(L)
L.append(55)# number 55 added to the end
print(L)
L.insert(3, 77)
print(L)# number 7 placed at index 3, right shifted elements

L.remove(40)
print(L)
L.pop()
print(L)

L.sort()
print("sorted ",L)

L.sort(reverse=True)
print("Reverse sorted ", L)

print(L[:3])

print(L[-3:])
avg = sum(L) / len(L)
L_above_avg = []

for x in L:
    if x > avg:
        L_above_avg.append(x)
print(L_above_avg)
```

```python
scores=tuple(L[:8])

a=max(scores)
b=scores.index(a)

c=min(scores)
d=scores.count(c)

print(a,b)
print(c,d)

# tuple is immutable so it cant be reversed in place
x=list(reversed(scores))
print(x)

n=int(input("score: "))
if n in scores:
 print(scores.index(n))
else:
 print("not present")

try:
 scores[0]=100
except TypeError as e:
 print(e)
 # tuple cant be changed like list

a,b,*c=scores
print(a,b,c)
```

```python
import random

random.seed(1024170220)

L=[]
for i in range(100):
 L.append(random.randint(100,900))

odd=[x for x in L if x%2!=0]
even=[x for x in L if x%2==0]

def prime(x):
 if x<2:
  return False
 for i in range(2,int(x**0.5)+1):
  if x%i==0:
   return False
 return True

p=[x for x in L if prime(x)]

print(odd)
print(len(odd))
print(even)
print(len(even))
print(p)
print(len(p))

m=max(L,key=L.count)
print(m,L.count(m))
```

```python
A=set(x*7 for x in L[:8])
B=set(x*9 for x in L[:8])

print(A)
print(B)

print(A.union(B))
print(A.intersection(B))

print(A.difference(B))
print(B.difference(A))
# difference gives values only in one set, symmetric difference gives values in either set but not both

print(A.symmetric_difference(B))

print(A.issubset(B))
print(B.issuperset(A))

x=int(input("x: "))
A.discard(x)
# discard doesnt give an error if x is not in the set like remove does
print(A)
```
